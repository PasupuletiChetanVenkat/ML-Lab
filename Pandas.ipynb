# **Column Addition**

```python
# AP24110010077

info = {
    'one': pd.Series(
        [1, 2, 3, 4, 5],
        index=['a', 'b', 'c', 'd', 'e']
    ),

    'two': pd.Series(
        [1, 2, 3, 4, 5, 6],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    )
}

df = pd.DataFrame(info)

print("Original DataFrame:")
print(df)

df['three'] = pd.Series(
    [20, 40, 60],
    index=['a', 'b', 'c']
)

print("\nAfter adding third column:")
print(df)

df['four'] = df['one'] + df['three']

print("\nAfter adding fourth column:")
print(df)
```

# **Addition of Rows**

```python
# AP24110010077

d = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

d2 = pd.DataFrame({
    'A': [7, 8, 9],
    'B': [10, 11, 12]
})

d = pd.concat([d, d2], ignore_index=True)

print(d)
```

# **Deletion of Rows**

```python
# AP24110010077

a_info = pd.DataFrame(
    [[4, 5], [6, 7]],
    columns=['x', 'y']
)

b_info = pd.DataFrame(
    [[8, 9], [10, 11]],
    columns=['x', 'y']
)

a_info = pd.concat([a_info, b_info], ignore_index=True)

print("Original DataFrame:")
print(a_info)

a_info = a_info.drop(0)

print("\nAfter deleting row 0:")
print(a_info)
```
