```python
sentence = "NumPy makes numerical computing fast and NumPy makes arrays powerful"

# Basic tokenization: lowercase + split on spaces
#AP24110010077
#The first step in any NLP pipeline is tokenization — splitting text into individual words.
tokens = sentence.lower().split()#AP24110010077
print(tokens)
```

```python
tokens_array = np.array(tokens)
vocabulary = np.unique(tokens_array)#AP24110010077
print("Vocabulary:", vocabulary)#A vocabulary is the set of all unique words across your text(s). We'll use np.unique() — the same function idea you used earlier for arrays of numbers, now applied to an array of words
print("Vocabulary size:", len(vocabulary))
```

```python
def bag_of_words_vector(tokens, vocabulary):
    # For each word in the vocabulary, count how many times it appears in tokens
    vector = np.array([np.sum(np.array(tokens) == word) for word in vocabulary])
    return vector#AP24110010077

vec = bag_of_words_vector(tokens, vocabulary)#AP24110010077
print(vocabulary)
print(vec)
```

```python
sentence_a = "machine learning uses numpy arrays"
sentence_b = "deep learning also uses numpy arrays heavily"

tokens_a = sentence_a.lower().split()
tokens_b = sentence_b.lower().split()#AP24110010077

# Build a shared vocabulary from both sentences combined
combined_vocab = np.unique(np.array(tokens_a + tokens_b))

vec_a = bag_of_words_vector(tokens_a, combined_vocab)#AP24110010077
vec_b = bag_of_words_vector(tokens_b, combined_vocab)

print("Vector A:", vec_a)#AP24110010077
print("Vector B:", vec_b)

# Cosine similarity using NumPy dot product and norm
cosine_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))
print("Cosine similarity:", cosine_sim)
```

```python
import numpy as np
import re

# Step 1: Take a paragraph
text = """
Python is a powerful programming language.
Python is easy to learn and Python is widely used.
Programming with Python is useful for data science.
"""

# Step 2: Tokenize the paragraph
tokens = re.findall(r'\b\w+\b', text.lower())#AP24110010077

print("Tokens:")
print(tokens)

# Step 3: Build the vocabulary
vocabulary = sorted(set(tokens))#AP24110010077

print("\nVocabulary:")
print(vocabulary)

# Step 4: Build the word-count vector
word_count_vector = np.array([tokens.count(word) for word in vocabulary])#AP24110010077

print("\nWord-count vector:")#AP24110010077
print(word_count_vector)

# Step 5: Find the indices of words sorted by frequency
sorted_indices = np.argsort(word_count_vector)

# Reverse the order to get highest frequency first
top_indices = sorted_indices[::-1][:3]

# Step 6: Print the top 3 most frequent words
print("\nTop 3 most frequent words:")

for index in top_indices:
    print(vocabulary[index], ":", word_count_vector[index])
```
